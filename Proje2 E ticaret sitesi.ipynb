```python
class Ürün():
    def __init__(self, isim, fiyat, stok_miktarı, kategori):
        self.isim = isim
        self.fiyat = fiyat
        self.stok_miktarı = stok_miktarı
        self.kategori = kategori 
        self.satıs_sayısı = 0
        print(f"Ürün eklendi: {isim}")

    def bilgi_goster(self):
        stok_durumu = "Stokta var" if self.stok_miktarı > 0 else "Stokta yok"
        print(f"\nÜrün: {self.isim}")
        print(f"Fiyat: {self.fiyat} TL")
        print(f"Stok: {self.stok_miktarı}")
        print(f"Kategori: {self.kategori}")
        print(f"Satış Sayısı: {self.satıs_sayısı}")
        print(f"Durum: {stok_durumu}")

    def fiyat_güncelle(self, yeni_fiyat):
        eski_fiyat = self.fiyat
        self.fiyat = yeni_fiyat
        print(f"{self.isim} fiyat güncellendi: {eski_fiyat} TL --> {self.fiyat} TL") # f-string kapatıldı.

    def stok_ekle(self, miktar): 
        self.stok_miktarı += miktar
        print(f"{self.isim} stok artırıldı: +{miktar} adet (Toplam: {self.stok_miktarı} adet)") # f-string kapatıldı.

    def satıs_yap(self, miktar):
        if self.stok_miktarı >= miktar:
            self.stok_miktarı -= miktar 
            self.satıs_sayısı += miktar 
            toplam_gelir = miktar * self.fiyat
            print(f"{miktar} adet {self.isim} satıldı. Gelir: {toplam_gelir} TL")
            print(f"Kalan stok: {self.stok_miktarı} adet")
            return toplam_gelir
        else:
            print(f"Yetersiz stok! İstenen miktar: {miktar} adet, Mevcut: {self.stok_miktarı} adet") # Print ifadesi düzeltildi.
            return 0

    def indirim_uygula(self, indirim_yüzdesi):
        eski_fiyat = self.fiyat
        
        self.fiyat = self.fiyat * (1 - indirim_yüzdesi / 100)
        print(f"{self.isim} ürününe %{indirim_yüzdesi} indirim uygulandı.")
        print(f"Eski fiyat: {eski_fiyat:.2f} TL --> Yeni fiyat: {self.fiyat:.2f} TL")


urun1 = Ürün("Laptop", 12000, 50, "Elektronik")
urun1.bilgi_goster()

urun1.fiyat_güncelle(11500)
urun1.bilgi_goster()

urun1.stok_ekle(10)
urun1.bilgi_goster()

urun1.satıs_yap(5)
urun1.bilgi_goster()

urun1.satıs_yap(60) 

urun1.indirim_uygula(10)
urun1.bilgi_goster()
```
